```python
df_geo["time"] = pd.to_datetime(df_geo["time"])

pairs = []

for i in range(len(df_geo)):
    current_time = df_geo.iloc[i]["time"]

    # Find only events occurring within the next 1 hour
    mask = (
        (df_geo["time"] > current_time) &
        (df_geo["time"] <= current_time + pd.Timedelta(hours=1))
    )

    nearby_time = df_geo.loc[mask]

    for j, row in nearby_time.iterrows():
        pairs.append({
            "earthquake_1": df_geo.iloc[i]["id"],
            "earthquake_2": row["id"],
            "time_1": current_time,
            "time_2": row["time"],
            "place_1": df_geo.iloc[i]["place"],
            "place_2": row["place"],
            "lat1": df_geo.iloc[i]["latitude"],
            "lon1": df_geo.iloc[i]["longitude"],
            "lat2": row["latitude"],
            "lon2": row["longitude"]
        })

print("Candidate pairs:", len(pairs))
```

```python
df_pairs = pd.DataFrame(pairs)

print("Total candidate pairs:", len(df_pairs))
```

```python
from numpy import radians, sin, cos, sqrt, arcsin

lat1 = radians(df_pairs["lat1"])
lat2 = radians(df_pairs["lat2"])

dlat = radians(df_pairs["lat2"] - df_pairs["lat1"])
dlon = radians(df_pairs["lon2"] - df_pairs["lon1"])

a = (
    sin(dlat / 2) ** 2
    + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2
)

df_pairs["distance_km"] = 6371 * 2 * arcsin(sqrt(a))
```

```python
df_query29 = df_pairs[
    df_pairs["distance_km"] <= 50
].copy()
```

```python
df_query29["time_difference_minutes"] = (
    pd.to_datetime(df_query29["time_2"])
    - pd.to_datetime(df_query29["time_1"])
).dt.total_seconds() / 60
```

```python
# ============================================
# GLOBAL SEISMIC TRENDS - 26 FEATURES
# Corrected Error-Free Version
# ============================================

import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import warnings

warnings.filterwarnings("ignore")


# ============================================
# SET VISUALIZATION STYLE
# ============================================

plt.style.use("seaborn-v0_8-darkgrid")
sns.set_palette("husl")


# ============================================
# 1. FETCH DATA FROM USGS API
# ============================================

def fetch_usgs_earthquake_data(years_back=5, min_magnitude=4.0):
    """
    Fetch earthquake data from USGS API
    and extract all specified features.
    """

    print(
        f"\n📥 Fetching earthquake data "
        f"(Magnitude >= {min_magnitude}) "
        f"for the last {years_back} years..."
    )

    url = "https://earthquake.usgs.gov/fdsnws/event/1/query"

    all_records = []

    current_year = datetime.now().year
    current_month = datetime.now().month
    start_year = current_year - years_back

    # --------------------------------------------
    # Fetch data year by year and month by month
    # --------------------------------------------

    for year in range(start_year, current_year + 1):

        print(f"\nFetching data for {year}...")

        for month in range(1, 13):

            # Skip future months in current year
            if year == current_year and month > current_month:
                continue

            start_date = f"{year}-{month:02d}-01"

            if month == 12:
                end_date = f"{year + 1}-01-01"
            else:
                end_date = f"{year}-{month + 1:02d}-01"

            params = {
                "format": "geojson",
                "starttime": start_date,
                "endtime": end_date,
                "minmagnitude": min_magnitude
            }

            try:

                response = requests.get(
                    url,
                    params=params,
                    timeout=30
                )

                if response.status_code != 200:

                    print(
                        f"  ⚠️ Failed for {start_date}: "
                        f"HTTP {response.status_code}"
                    )

                    continue

                data = response.json()

                features = data.get("features", [])

                print(
                    f"  {start_date}: "
                    f"{len(features)} earthquakes"
                )

                # --------------------------------------------
                # Extract earthquake records
                # --------------------------------------------

                for feature in features:

                    properties = feature.get("properties", {})

                    geometry = feature.get("geometry", {})

                    coordinates = geometry.get("coordinates")

                    # Make sure coordinates exist
                    if not coordinates:
                        latitude = None
                        longitude = None
                        depth_km = None

                    else:

                        longitude = (
                            coordinates[0]
                            if len(coordinates) > 0
                            else None
                        )

                        latitude = (
                            coordinates[1]
                            if len(coordinates) > 1
                            else None
                        )

                        depth_km = (
                            coordinates[2]
                            if len(coordinates) > 2
                            else None
                        )

                    record = {

                        "id": feature.get("id"),

                        "time": (
                            pd.to_datetime(
                                properties.get("time"),
                                unit="ms"
                            )
                            if properties.get("time")
                            else None
                        ),

                        "updated": (
                            pd.to_datetime(
                                properties.get("updated"),
                                unit="ms"
                            )
                            if properties.get("updated")
                            else None
                        ),

                        "latitude": latitude,

                        "longitude": longitude,

                        "depth_km": depth_km,

                        "mag": properties.get("mag"),

                        "magType": properties.get("magType"),

                        "place": properties.get("place"),

                        "status": properties.get("status"),

                        "tsunami": properties.get("tsunami"),

                        "sig": properties.get("sig"),

                        "net": properties.get("net"),

                        "nst": properties.get("nst"),

                        "dmin": properties.get("dmin"),

                        "rms": properties.get("rms"),

                        "gap": properties.get("gap"),

                        "magError": properties.get("magError"),

                        "depthError": properties.get("depthError"),

                        "magNst": properties.get("magNst"),

                        "locationSource": properties.get(
                            "locationSource"
                        ),

                        "magSource": properties.get(
                            "magSource"
                        ),

                        "types": properties.get("types"),

                        "ids": properties.get("ids"),

                        "sources": properties.get("sources"),

                        "type": properties.get("type")
                    }

                    all_records.append(record)

            except Exception as e:

                print(
                    f"  ⚠️ Error fetching "
                    f"{start_date}: {e}"
                )

                continue

    # ============================================
    # CREATE DATAFRAME
    # ============================================

    df = pd.DataFrame(all_records)

    # Check whether data was received
    if df.empty:

        print("\n❌ No earthquake data was retrieved.")

        return df

    # --------------------------------------------
    # Remove rows missing crucial fields
    # --------------------------------------------

    required_columns = [
        "mag",
        "latitude",
        "longitude"
    ]

    df = df.dropna(
        subset=required_columns
    ).copy()

    # ============================================
    # DATA FETCH COMPLETE
    # ============================================

    print("\n" + "=" * 60)

    print("✅ DATA FETCH COMPLETE!")

    print("=" * 60)

    print(
        f"Dataset loaded: "
        f"{df.shape[0]} rows × "
        f"{df.shape[1]} columns"
    )

    print("\nFeatures successfully extracted:")

    print(
        ", ".join(
            df.columns.tolist()
        )
    )

    print("\n" + "=" * 60)

    print("FEATURES SUCCESSFULLY EXTRACTED")

    print("=" * 60)

    return df


# ============================================
# 2. KEY METRICS CALCULATION
# ============================================

def calculate_seismic_metrics(
    df,
    mag_column="mag",
    depth_column="depth_km"
):
    """
    Calculate key seismic metrics.
    """

    print("\n" + "=" * 50)

    print("SEISMIC DATA ANALYSIS METRICS")

    print("=" * 50)

    metrics = {}

    # --------------------------------------------
    # Basic metrics
    # --------------------------------------------

    metrics["Total Earthquakes"] = len(df)

    metrics["Mean Magnitude"] = (
        df[mag_column].mean()
    )

    metrics["Max Magnitude"] = (
        df[mag_column].max()
    )

    # --------------------------------------------
    # Tsunami metric
    # --------------------------------------------

    if "tsunami" in df.columns:

        metrics["Total Tsunamis Generated"] = (
            df["tsunami"]
            .fillna(0)
            .sum()
        )

    else:

        metrics["Total Tsunamis Generated"] = 0

    # --------------------------------------------
    # Significance
    # --------------------------------------------

    if "sig" in df.columns:

        metrics["Avg Impact Significance (sig)"] = (
            df["sig"].mean()
        )

    # --------------------------------------------
    # Depth metrics
    # --------------------------------------------

    if depth_column in df.columns:

        metrics["Mean Depth (km)"] = (
            df[depth_column].mean()
        )

        metrics["Max Depth (km)"] = (
            df[depth_column].max()
        )

    # ============================================
    # MAGNITUDE CATEGORIES
    # ============================================

    def categorize_magnitude(mag):

        if mag < 3.0:
            return "Micro"

        elif mag < 4.0:
            return "Minor"

        elif mag < 5.0:
            return "Light"

        elif mag < 6.0:
            return "Moderate"

        elif mag < 7.0:
            return "Strong"

        elif mag < 8.0:
            return "Major"

        else:
            return "Great"

    df["magnitude_category"] = (
        df[mag_column]
        .apply(categorize_magnitude)
    )

    # ============================================
    # PRINT METRICS
    # ============================================

    for key, value in metrics.items():

        if isinstance(value, (float, np.floating)):

            print(
                f"{key}: {value:.2f}"
            )

        else:

            print(
                f"{key}: {value}"
            )

    return metrics, df


# ============================================
# 3. TEMPORAL TREND ANALYSIS
# ============================================

def analyze_temporal_trends(
    df,
    date_column="time"
):
    """
    Analyze earthquake frequency over time.
    """

    print("\n" + "=" * 50)

    print("TEMPORAL TREND ANALYSIS")

    print("=" * 50)

    # --------------------------------------------
    # Make sure time is datetime
    # --------------------------------------------

    df[date_column] = pd.to_datetime(
        df[date_column],
        errors="coerce"
    )

    # --------------------------------------------
    # Create year and month columns
    # --------------------------------------------

    df["year"] = (
        df[date_column]
        .dt.year
    )

    df["month"] = (
        df[date_column]
        .dt.month
    )

    # --------------------------------------------
    # Yearly earthquake count
    # --------------------------------------------

    yearly_trend = (
        df.groupby("year")
        .size()
    )

    print("\nYearly earthquake counts:")

    print(yearly_trend)

    return {
        "yearly": yearly_trend
    }, df


# ============================================
# 4. SPATIAL ANALYSIS
# ============================================

def analyze_spatial_distribution(
    df,
    lat_col="latitude",
    lon_col="longitude"
):
    """
    Analyze geographic distribution.
    """

    print("\n" + "=" * 50)

    print("SPATIAL DISTRIBUTION ANALYSIS")

    print("=" * 50)

    spatial_stats = {

        "Mean Latitude": (
            df[lat_col].mean()
        ),

        "Mean Longitude": (
            df[lon_col].mean()
        )
    }

    print(
        f"Average Epicenter: "
        f"{spatial_stats['Mean Latitude']:.2f} Lat, "
        f"{spatial_stats['Mean Longitude']:.2f} Lon"
    )

    return spatial_stats, df


# ============================================
# 5. VISUALIZATION FUNCTIONS
# ============================================

def create_seismic_visualizations(
    df,
    mag_column="mag",
    depth_column="depth_km"
):
    """
    Create comprehensive visualizations.
    """

    plt.rcParams.update({

        "font.size": 18,

        "axes.titlesize": 24,

        "axes.labelsize": 18,

        "xtick.labelsize": 15,

        "ytick.labelsize": 15
    })

    # ============================================
    # Create figure
    # ============================================

    fig, axes = plt.subplots(
        2,
        3,
        figsize=(24, 14)
    )

    # ============================================
    # 1. MAGNITUDE DISTRIBUTION
    # ============================================

    axes[0, 0].hist(
        df[mag_column],
        bins=30,
        edgecolor="black",
        alpha=0.7
    )

    axes[0, 0].set_xlabel(
        "Magnitude (mag)"
    )

    axes[0, 0].set_ylabel(
        "Frequency"
    )

    axes[0, 0].set_title(
        "Earthquake Magnitude Distribution"
    )

    # ============================================
    # 2. MAGNITUDE VS DEPTH
    # ============================================

    if depth_column in df.columns:

        scatter = axes[0, 1].scatter(

            df[mag_column],

            df[depth_column],

            alpha=0.3,

            c=df[mag_column],

            cmap="viridis"
        )

        axes[0, 1].set_xlabel(
            "Magnitude"
        )

        axes[0, 1].set_ylabel(
            "Depth (km)"
        )

        axes[0, 1].set_title(
            "Magnitude vs Depth Relationship"
        )

        axes[0, 1].invert_yaxis()

        plt.colorbar(
            scatter,
            ax=axes[0, 1],
            label="Magnitude"
        )

    # ============================================
    # 3. YEARLY TREND
    # ============================================

    if "year" in df.columns:

        yearly_counts = (
            df.groupby("year")
            .size()
        )

        axes[0, 2].plot(

            yearly_counts.index,

            yearly_counts.values,

            marker="o",

            linewidth=2,

            color="coral"
        )

        axes[0, 2].set_xlabel(
            "Year"
        )

        axes[0, 2].set_ylabel(
            "Total Earthquakes"
        )

        axes[0, 2].set_title(
            "Yearly Earthquake Frequency"
        )

        axes[0, 2].set_xticks(
            yearly_counts.index
        )

    # ============================================
    # 4. SIGNIFICANCE VS MAGNITUDE
    # ============================================

    if "sig" in df.columns:

        axes[1, 0].scatter(

            df[mag_column],

            df["sig"],

            alpha=0.4,

            color="purple"
        )

        axes[1, 0].set_xlabel(
            "Magnitude"
        )

        axes[1, 0].set_ylabel(
            "Significance Metric (sig)"
        )

        axes[1, 0].set_title(
            "Earthquake Impact Significance by Magnitude"
        )

    # ============================================
    # 5. SPATIAL DISTRIBUTION
    # ============================================

    scatter_map = axes[1, 1].scatter(

        df["longitude"],

        df["latitude"],

        c=df[mag_column],

        s=np.maximum(
            (df[mag_column] - 2) ** 3,
            5
        ),

        alpha=0.5,

        cmap="Reds"
    )

    axes[1, 1].set_xlabel(
        "Longitude"
    )

    axes[1, 1].set_ylabel(
        "Latitude"
    )

    axes[1, 1].set_title(
        "Global Epicenter Distribution"
    )

    plt.colorbar(
        scatter_map,
        ax=axes[1, 1],
        label="Magnitude"
    )

    # ============================================
    # 6. TSUNAMI OCCURRENCES
    # ============================================

    if "tsunami" in df.columns:

        tsunami_counts = (

            df[df["tsunami"] == 1]

            .groupby(
                "magnitude_category"
            )

            .size()
        )

        if not tsunami_counts.empty:

            axes[1, 2].bar(

                tsunami_counts.index,

                tsunami_counts.values,

                color="teal"
            )

            axes[1, 2].set_xlabel(
                "Magnitude Category"
            )

            axes[1, 2].set_ylabel(
                "Number of Tsunamis"
            )

            axes[1, 2].set_title(
                "Tsunami Generation by Magnitude Category"
            )

        else:

            axes[1, 2].text(

                0.5,
                0.5,

                "No tsunami events found",

                ha="center",

                va="center",

                fontsize=16
            )

            axes[1, 2].set_title(
                "Tsunami Generation by Magnitude Category"
            )

    # ============================================
    # FINALIZE FIGURE
    # ============================================

    plt.tight_layout()

    output_image = (
        "global_seismic_trends_26_features.png"
    )

    plt.savefig(
        output_image,
        dpi=300,
        bbox_inches="tight"
    )

    print(
        f"\n✅ Visualizations saved as "
        f"'{output_image}'"
    )

    plt.show()


# ============================================
# 6. MAIN EXECUTION PIPELINE
# ============================================

if __name__ == "__main__":

    print(
        "🌍 GLOBAL SEISMIC TRENDS: "
        "26-FEATURE DATASET"
    )

    print("=" * 60)

    # ============================================
    # STEP 1: FETCH DATA
    # ============================================

    # 4.0 = faster testing
    # Use 3.0 for a larger dataset

    df_raw = fetch_usgs_earthquake_data(

        years_back=5,

        min_magnitude=4.0
    )

    # ============================================
    # CHECK WHETHER DATA WAS FETCHED
    # ============================================

    if df_raw.empty:

        print(
            "\n❌ No data available. "
            "Pipeline stopped."
        )

    else:

        # ========================================
        # STEP 2: CALCULATE METRICS
        # ========================================

        metrics, df = (
            calculate_seismic_metrics(
                df_raw
            )
        )

        # ========================================
        # STEP 3: NUMBER THE COLUMNS
        # ========================================

        print("\n" + "=" * 60)

        print("NUMBERED COLUMN LIST")

        print("=" * 60)

        for i, column in enumerate(
            df.columns,
            start=1
        ):

            print(
                f"{i}. {column}"
            )

        # ========================================
        # STEP 4: TEMPORAL ANALYSIS
        # ========================================

        temporal_results, df = (
            analyze_temporal_trends(
                df
            )
        )

        # ========================================
        # STEP 5: SPATIAL ANALYSIS
        # ========================================

        spatial_stats, df = (
            analyze_spatial_distribution(
                df
            )
        )

        # ========================================
        # STEP 6: DATA PROFILING
        # ========================================

        print("\n" + "=" * 60)

        print("DATASET INFORMATION")

        print("=" * 60)

        # ----------------------------------------
        # Columns
        # ----------------------------------------

        print("\nColumns:")

        print(
            df.columns.tolist()
        )

        # ----------------------------------------
        # Data types
        # ----------------------------------------

        print("\nData types:")

        print(
            df.dtypes
        )

        # ----------------------------------------
        # First 5 rows
        # ----------------------------------------

        print("\nFirst 5 rows:")

        print(
            df.head()
        )

        # ----------------------------------------
        # Missing values
        # ----------------------------------------

        print("\nMissing values:")

        print(
            df.isnull().sum()
        )

        # ----------------------------------------
        # Duplicate IDs
        # ----------------------------------------

        print("\nDuplicate IDs:")

        if "id" in df.columns:

            print(
                df["id"]
                .duplicated()
                .sum()
            )

        # ========================================
        # STEP 7: VISUALIZATIONS
        # ========================================

        print(
            "\n🎨 Generating visualizations..."
        )

        create_seismic_visualizations(
            df
        )

        # ========================================
        # STEP 8: SAVE DATA AS CSV
        # ========================================

        csv_name = (
            "usgs_earthquakes_26_features.csv"
        )

        df.to_csv(
            csv_name,
            index=False
        )

        print(
            f"\n💾 Data saved locally to "
            f"'{csv_name}'"
        )

        # ========================================
        # STEP 9: CREATE SUMMARY TABLE
        # ========================================

        summary = pd.DataFrame({

            "Column": df.columns,

            "Data Type": (
                df.dtypes.astype(str)
            ),

            "Missing Values": (
                df.isnull()
                .sum()
                .values
            ),

            "Unique Values": (
                df.nunique()
                .values
            )
        })

        print("\n" + "=" * 60)

        print("DATASET SUMMARY")

        print("=" * 60)

        print(summary)

        # ========================================
        # DISPLAY SUMMARY
        # ========================================

        try:

            display(summary)

        except NameError:

            print(
                "\nSummary displayed above."
            )

        # ========================================
        # COMPLETION MESSAGE
        # ========================================

        print("\n" + "=" * 60)

        print(
            "✅ PIPELINE EXECUTION "
            "COMPLETED SUCCESSFULLY! 🚀"
        )

        print("=" * 60)
```
